```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1. Pandas: iris.csv einlesen und die ersten Zeilen anzeigen
print("--- 1. Pandas: Daten laden und anzeigen ---")
df = pd.read_csv("/workspaces/Python/Test/iris.csv")
print(df.head(), "\n")
```

```python
# 2. NumPy: Statistiken über numerische Spalten berechnen
print("--- 2. NumPy: Berechnungen auf Spalten ---")
sepal_lengths = df["sepal_length"].to_numpy()
mean_length = np.mean(sepal_lengths)
std_length = np.std(sepal_lengths)

print(f"Sepal Length - Mittelwert: {mean_length:.2f}")
print(f"Sepal Length - Standardabweichung: {std_length:.2f}\n")
```

```python
# 3. Matplotlib: Einen Scatter Plot der Iris-Daten erstellen und anzeigen
print("--- 3. Matplotlib: Diagramm erstellen ---")
plt.figure(figsize=(8, 6))

# Für jede Spezies eine andere Farbe verwenden
for species, group in df.groupby("species"):
    plt.scatter(group["sepal_length"], group["sepal_width"], label=species)

plt.title("Iris Dataset - Sepal Length vs Width")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.legend()
plt.grid(True)

# Diagramm anzeigen
plt.show()
```
